 INTRODUCCION
 Este programa permite registrar visitantes del museo, calcular el precio de entrada según su edad y aplicar el descuento correspondiente. Al finalizar, muestra el número de visitantes cobrados, el total de descuentos y el total a pagar.

 **PARTE B:**   
Parte B — El programa (45 pts)
En `semana2/taquilla_museo.ipynb`. Tu programa debe:

1. Usar **constantes** al inicio para los precios y los descuentos. Por ejemplo: `PRECIO_ADULTO = 45`.
2. Usar un ciclo **`while`** que atienda visitantes hasta que se escriba `fin`.
3. Usar **`break`** para cerrar la taquilla al recibir `fin`.
4. Usar **`continue`** para saltar una edad no válida sin cobrarla.
5. Calcular el precio y el descuento con **`if / elif / else`** siguiendo tu tabla de la Parte A.
6. Preguntar el tipo de visitante **solo** a los mayores de 18, con este menú: `1) Estudiante  2) Profesor  3) Ninguno`.
7. Mostrar el **detalle de cada visitante** en una línea: número, edad, tipo, precio base, descuento y cuánto paga.
8. Al cerrar, mostrar el **corte de caja**: visitantes cobrados, total en descuentos y total a pagar. Necesitas **acumuladores** y un **contador**.
9. Mostrar todos los montos con 2 decimales.

```python
visitantes_cobrados = 0
total_descuentos = 0.00
total_a_pagar = 0.00

print("=== TAQUILLA DEL MUSEO ===")

PRECIO_MENOR_3 = 0.00
PRECIO_MENOR_18 = 30.00
PRECIO_ADULTO = 45.00

DESCUENTO_ADULTO_MAYOR = 0.12
DESCUENTO_ESTUDIANTE = 0.10
DESCUENTO_PROFESOR = 0.10
while True:
    entrada = input(
        "Edad del visitante (o 'fin' para cerrar la taquilla): "
    ).strip().lower()
    if entrada == "fin":
        break
    try:
        edad = int(entrada)
    except ValueError:
        print("Entrada inválida. Escribe una edad entera o 'fin'.")
        continue
    if edad <=0 or edad > 110:
        print("Edad no válida. No se cobra; ingresa al siguiente visitante.")
        continue
    descuento = 0.00
    if edad < 3:
        precio_base = PRECIO_MENOR_3
        tipo = "Menor de 3 años"
    elif edad < 18:
        precio_base = PRECIO_MENOR_18
        tipo = "Menor de edad"
    else:
        precio_base = PRECIO_ADULTO
        opcion = input(
            "Tipo: 1) Estudiante  2) Profesor  3) Ninguno: "
        ).strip()

        if opcion == "1":
            tipo = "Estudiante"
        elif opcion == "2":
            tipo = "Profesor"
        elif opcion == "3":
            tipo = "Ninguno"
        else:
            print("Opción inválida. Se registrará como Ninguno.")
            tipo = "Ninguno"
        if edad >= 60:
            descuento = precio_base * DESCUENTO_ADULTO_MAYOR
            tipo = "Adulto mayor (" + tipo + ")"

        elif tipo == "Estudiante":
            descuento = precio_base * DESCUENTO_ESTUDIANTE

        elif tipo == "Profesor":
            descuento = precio_base * DESCUENTO_PROFESOR
    a_pagar = precio_base - descuento
    visitantes_cobrados += 1
    total_descuentos += descuento
    total_a_pagar += a_pagar
    print(
        f"Visitante {visitantes_cobrados} | "
        f"Edad: {edad} | {tipo} | "
        f"Base: ${precio_base:.2f} | "
        f"Descuento: ${descuento:.2f} | "
        f"A pagar: ${a_pagar:.2f}")

print("\n=== CORTE DE CAJA ===")
print(f"Visitantes cobrados: {visitantes_cobrados}")
print(f"Total en descuentos: ${total_descuentos:.2f}")
print(f"TOTAL A PAGAR: ${total_a_pagar:.2f}")
```
