```python
def sumar_tupla(tupla):
    total = 0
    for numero in tupla:
        total += numero
    return total


def seccion_tuplas():
    numeros = (5, 10, 15, 20, 25)

    print("\n--- SECCIÓN DE TUPLAS ---")
    print("Tupla original:", numeros)
    print("Tercer elemento:", numeros[2])

    numero1 = int(input("Ingresa el primer número: "))
    numero2 = int(input("Ingresa el segundo número: "))

    nueva_tupla = numeros + (numero1, numero2)

    lista_numeros = list(nueva_tupla)
    lista_numeros.sort()

    print("Nueva tupla:", nueva_tupla)
    print("Lista ordenada:", lista_numeros)
    print("Suma de la nueva tupla:", sumar_tupla(nueva_tupla))


def buscar_contacto(contactos, nombre):
    return contactos.get(nombre)


def seccion_diccionarios():
    contactos = {
        "Ana": "4421234567",
        "Luis": "4422345678",
        "Maria": "4423456789"
    }

    print("\n--- SECCIÓN DE DICCIONARIOS ---")

    nombre_nuevo = input("Ingresa el nombre del nuevo contacto: ")
    telefono_nuevo = input("Ingresa su número de teléfono: ")

    contactos[nombre_nuevo] = telefono_nuevo

    print("\nNombres registrados:")
    for nombre in contactos:
        print(nombre)

    nombre_buscar = input("\n¿Qué contacto deseas buscar? ")

    telefono = buscar_contacto(contactos, nombre_buscar)

    if telefono is not None:
        print("Teléfono de", nombre_buscar, ":", telefono)
    else:
        print(nombre_buscar, "no está en tus contactos")


def seccion_excepciones():
    print("\n--- SECCIÓN DE EXCEPCIONES ---")

    try:
        numero1 = int(input("Ingresa el primer número entero: "))
        numero2 = int(input("Ingresa el segundo número entero: "))

        print("Suma:", numero1 + numero2)

        division = numero1 / numero2
        print("División:", division)

    except ValueError:
        print("Error: debes ingresar números enteros y no dejar espacios vacíos.")

    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")


def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)


def seccion_strings():
    print("\n--- SECCIÓN DE STRINGS ---")

    mensaje = "La programación permite crear soluciones útiles"

    print("Mensaje original:", mensaje)
    print("Longitud del mensaje:", len(mensaje))
    print("Mensaje en mayúsculas:", mensaje.upper())

    mensaje_modificado = mensaje.replace("útiles", "creativas")

    print("Mensaje modificado:", mensaje_modificado)
    print("Cantidad de palabras:", contar_palabras(mensaje))


def mostrar_menu():
    while True:
        print("\n========== MENÚ MODULAR ==========")
        print("1) Tuplas")
        print("2) Diccionarios")
        print("3) Excepciones")
        print("4) Strings")
        print("5) Salir")

        opcion = input("Elige una opción: ")

        if opcion == "1":
            seccion_tuplas()

        elif opcion == "2":
            seccion_diccionarios()

        elif opcion == "3":
            seccion_excepciones()

        elif opcion == "4":
            seccion_strings()

        elif opcion == "5":
            print("Gracias por utilizar el programa. ¡Hasta pronto!")
            break

        else:
            print("Opción inválida. Intenta nuevamente.")


mostrar_menu()
```
