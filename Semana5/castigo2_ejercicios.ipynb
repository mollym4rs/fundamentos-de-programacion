# Ejercicio 1
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

Crea la función resumen(km) que regrese una tupla con: total de km, cuántos días sí se rodó (km > 0), el promedio de km solo de los días rodados (redondeado a 1 decimal) y el índice del día más largo. Desempaca el resultado y muéstralo. Usa: función con return de una tupla, for, if, desempaquetar. Salida esperada:

Total: 660 km
Días con rodada: 5
Promedio por día rodado: 132.0 km
Día más largo: Jueves (210 km)

````python
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(kilometros):
    total = sum(kilometros)
    dias_rodados = 0
    suma_rodados = 0
    max_km = -1
    idx_max = 0
    
    for i in range(len(kilometros)):
        k = kilometros[i]
        if k > 0:
            dias_rodados += 1
            suma_rodados += k
        if k > max_km:
            max_km = k
            idx_max = i
            
    promedio = round(suma_rodados / dias_rodados, 1) if dias_rodados > 0 else 0.0
    return (total, dias_rodados, promedio, idx_max)

total, dias_rodados, promedio, idx_max = resumen(km)
print(f"Total: {total} km")
print(f"Días con rodada: {dias_rodados}")
print(f"Promedio por día rodado: {promedio} km")
print(f"Día más largo: {dias[idx_max]} ({km[idx_max]} km)")
````

# Ejercicio 7 — Boleta del grupo

Crea promedio(lista) que sume con un for y regrese el promedio redondeado a 1 decimal. Guarda los promedios en un diccionario nuevo, muestra si cada alumno aprueba (70 o más) y quién tiene el mejor promedio. Usa: diccionario con listas como valores, función con return, round(), items(). Salida esperada:

Ana: 84.3 - Aprobado
Luis: 65.7 - Reprobado

````python
notas = {"Ana": [90, 85, 78], "Luis": [60, 72, 65], "Mía": [100, 95, 98], "Beto": [70, 68, 74]}

def promedio(lista):
    suma = 0
    for n in lista:
        suma += n
    return round(suma / len(lista), 1)

promedios = {}
mejor_promedio = -1
mejor_alumno = ""

for alumno, califs in notas.items():
    p = promedio(califs)
    promedios[alumno] = p
    estado = "Aprobado" if p >= 70 else "Reprobado"
    print(f"{alumno}: {p} - {estado}")
    if p > mejor_promedio:
        mejor_promedio = p
        mejor_alumno = alumno

print(f"Mejor promedio: {mejor_alumno}")
````

abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

Crea expandir(mensaje) que pase el mensaje a minúsculas, lo corte en palabras y cambie cada abreviatura por su significado (las palabras que no están se quedan igual). Debe regresar el mensaje traducido. Usa: lower(), split(), get() con valor por defecto, lista, return. 💡 Pista: " ".join(lista) une una lista de palabras en un solo texto separado por espacios. Prueba con: "Q onda tmb voy al concierto xq tocan bn" y "tqm Mamá". Salida esperada:

que onda también voy al concierto porque tocan bien
te quiero mucho mamá

````python
abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    palabras = mensaje.lower().split()
    traducidas = []
    for p in palabras:
        traducidas.append(abrev.get(p, p))
    return " ".join(traducidas)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
````

Atrapa texto (`ValueError`) y posiciones que no existen (`IndexError`); si el asiento está ocupado, avisa. Repite hasta lograr **una** reserva y muestra la sala sin corchetes.

```

¡Reservado!
X L L
L X X
X X L

````python
sala = [["X", "L", "L"], ["L", "X", "L"], ["X", "X", "L"]]

def reservar(s):
    # Simulación de entradas de prueba: "uno" -> 0, 5 -> 0, 0 -> 1, 2
    intentos = [("uno", 0), (5, 0), (0, 1), (2, 2)]
    
    for f, c in intentos:
        try:
            fila = int(f)
            col = int(c)
            if s[fila][col] == "X":
                print("Ocupado, elige otro")
                continue
            else:
                s[fila][col] = "X"
                print("¡Reservado!")
                return True
        except ValueError:
            print("Escribe números")
        except IndexError:
            print("Ese asiento no existe")
    return False

reservar(sala)
for fila in sala:
    for val in fila:
        print(val, end=" ")
    print()
````

Crea `aplicar(total, codigo)` que limpie el código (`strip()` y `upper()`), busque el porcentaje en el diccionario y **regrese** el total con descuento. ```

Cupón no válido

````python
cupones = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}

def aplicar(total, codigo):
    try:
        limpio = codigo.strip().upper()
        descuento_porcentaje = cupones[limpio]
        nuevo_total = total * (1 - descuento_porcentaje / 100)
        return nuevo_total
    except KeyError:
        print("Cupón no válido")
        return total

codigos_prueba = [" moto20 ", "GRATIS", "metal10"]
total_base = 1200

for cod in codigos_prueba:
    res = aplicar(total_base, cod)
    print(f"'{cod}' → ${res}")
````

# Ejercicio 16 — ¿Tu contraseña aguanta?
Crea `evaluar(pw)` que **regrese una lista** con las fallas encontradas: menos de 8 caracteres, sin ningún número, sin ninguna mayúscula o con espacios. Si la lista está vacía, la contraseña es segura. Prueba con `"metal"`, `"Harley2008"`, `"road glide 08"` y `"KLR650moto"`.
**Usa:** `len()`, `for` sobre un string, `in`, lista de fallas, función con `return`.
💡 Pista: `"7".isdigit()` da `True` y `"A".isupper()` da `True`.
**Salida esperada:**
```
metal: ['mínimo 8 caracteres', 'falta un número', 'falta una mayúscula']
Harley2008: segura
road glide 08: ['falta una mayúscula', 'sin espacios']
KLR650moto: segura

````python
def evaluar(pw):
    fallas = []
    if len(pw) < 8:
        fallas.append("mínimo 8 caracteres")
    
    tiene_num = False
    tiene_may = False
    tiene_esp = False
    
    for c in pw:
        if c.isdigit():
            tiene_num = True
        if c.isupper():
            tiene_may = True
        if c == " ":
            tiene_esp = True
            
    if not tiene_num:
        fallas.append("falta un número")
    if not tiene_may:
        fallas.append("falta una mayúscula")
    if tiene_esp:
        fallas.append("sin espacios")
        
    return fallas if fallas else "segura"

pruebas = ["metal", "Harley2008", "road glide 08", "KLR650moto"]
for p in pruebas:
    print(f"{p}: {evaluar(p)}")
````

# Ejercicio 18 — Contador de vocales
```python
frase = "el rock nunca va a morir en esta ciudad"
```
Crea `vocales(palabra)` que cuente sus vocales con un `for` y regrese el número. Construye un diccionario palabra → vocales y di cuál tiene más.
**Usa:** `split()`, `for` sobre un string, `in "aeiou"`, diccionario, función con `return`.
**Salida esperada:**
```
{'el': 1, 'rock': 1, 'nunca': 2, 'va': 1, 'a': 1, 'morir': 2, 'en': 1, 'esta': 2, 'ciudad': 3}
Más vocales: ciudad (3)

````python
frase = "el rock nunca va a morir en esta ciudad"

def vocales(palabra):
    contador = 0
    for c in palabra:
        if c in "aeiou":
            contador += 1
    return contador

palabras = frase.split()
resultado_dic = {}
max_v = -1
mas_vocales = ""

for p in palabras:
    v = vocales(p)
    resultado_dic[p] = v
    if v > max_v:
        max_v = v
        mas_vocales = p

print(resultado_dic)
print(f"Más vocales: {mas_vocales} ({max_v})")
````

prohibidas = ["tonto", "menso", "baboso"]
```
Crea `censurar(frase)` que cambie cada palabra prohibida (sin importar mayúsculas) por asteriscos del mismo largo y **regrese una tupla**: (frase censurada, cuántas censuró). Prueba con `"no seas menso y pásame el código"`, `"Tonto el que no haga commit"` y `"todo bien profe"`.
**Usa:** `split()`, `for` con `range(len(...))`, `lower()`, `in`, `"*" * len(...)`, `return` de una tupla, desempaquetar.
**Salida esperada:**
```
no seas ***** y pásame el código  (1 censuradas)
***** el que no haga commit  (1 censuradas)
todo bien profe  (0 censuradas)

````python
prohibidas = ["tonto", "menso", "baboso"]

def censurar(frase):
    palabras = frase.split()
    censuradas_cnt = 0
    
    for i in range(len(palabras)):
        palabra_limpia = palabras[i].lower()
        if palabra_limpia in prohibidas:
            palabras[i] = "*" * len(palabras[i])
            censuradas_cnt += 1
            
    return (" ".join(palabras), censuradas_cnt)

textos = ["no seas menso y pásame el código", "Tonto el que no haga commit", "todo bien profe"]
for t in textos:
    frase_c, cant = censurar(t)
    print(f"{frase_c}  ({cant} censuradas)")
````

# Ejercicio 21 — Estacionamiento del foro
```python
autos = [("ABC123", 9, 11), ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]
```
Cada tupla es (placa, hora de entrada, hora de salida). Tarifa: $20 la primera hora (o fracción, incluso 0 horas) y $15 cada hora adicional. Crea `cobro(entrada, salida)` con `return`, guarda los cobros en un diccionario placa → cobro y muestra el total del día recorriendo `values()`.
**Usa:** lista de tuplas, desempaquetar, función con `return`, `if`, diccionario, `values()`.
**Salida esperada:**
```
ABC123: 2 h → $35
XYZ789: 7 h → $110
MOT650: 1 h → $20
QRO001: 0 h → $20
Total del día: $185

````python
autos = [("ABC123", 9, 11), ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]

def cobro(entrada, salida):
    horas = salida - entrada
    if horas <= 0:
        total = 20
        horas = 0
    else:
        total = 20 + (horas - 1) * 15
    return horas, total

cobros_dict = {}
for placa, ent, sal in autos:
    h, total = cobro(ent, sal)
    cobros_dict[placa] = total
    print(f"{placa}: {h} h → ${total}")

total_dia = sum(cobros_dict.values())
print(f"Total del día: ${total_dia}")
````

```python

movs = ["Ana:-150", "Luis:-250", "Mía:+100", "Pedro:-50", "Ana:-cien", "Mía:-380", "Luis:-200"]
```
Cada movimiento es `"nombre:monto"`. Sepáralo con `split(":")` y convierte el monto con `int()`. Si el saldo quedaría negativo, rechaza el movimiento sin aplicarlo. Atrapa los montos inválidos (`ValueError`) y las personas sin pulsera (`KeyError`), y cuenta esos movimientos con error.
**Usa:** `split()`, `try` con dos `except`, diccionario, `if/else`, contador.
**Salida esperada:**
```
Luis: saldo insuficiente
Pedro no tiene pulsera
Monto inválido en 'Ana:-cien'
{'Ana': 350, 'Luis': 0, 'Mía': 20}
Movimientos con error: 2

````python
saldos = {"Ana": 500, "Luis": 200, "Mía": 300}
movs = ["Ana:-150", "Luis:-250", "Mía:+100", "Pedro:-50", "Ana:-cien", "Mía:-380", "Luis:-200"]

errores = 0

for mov in movs:
    try:
        nombre, monto_str = mov.split(":")
        monto = int(monto_str)
        
        if nombre not in saldos:
            raise KeyError(f"{nombre} no tiene pulsera")
            
        if saldos[nombre] + monto < 0:
            print(f"{nombre}: saldo insuficiente")
            continue
            
        saldos[nombre] += monto
        
    except ValueError:
        print(f"Monto inválido en '{mov}'")
        errores += 1
    except KeyError as e:
        print(e)
        errores += 1

print(saldos)
print(f"Movimientos con error: {errores}")
````

**Usa:** tupla + matriz, `for` anidados, construir una lista por columna, función con `return` reutilizada.
**Salida esperada:**
```
Ana: 86.25
Luis: 68.25
Mía: 97.5
Beto: 72.5
Parcial 1: 77.5
Parcial 2: 82.0

````python
nombres = ("Ana", "Luis", "Mía", "Beto")
cal = [[85, 90, 78, 92], [60, 75, 70, 68], [95, 98, 100, 97], [70, 65, 80, 75]]

def promedio(lista):
    return sum(lista) / len(lista)

mejor_promedio = -1
mejor_alumno = ""

for i in range(len(nombres)):
    p = promedio(cal[i])
    print(f"{nombres[i]}: {p}")
    if p > mejor_promedio:
        mejor_promedio = p
        mejor_alumno = nombres[i]

for j in range(4):
    parcial_col = []
    for i in range(len(cal)):
        parcial_col.append(cal[i][j])
    print(f"Parcial {j+1}: {promedio(parcial_col)}")

print(f"Mejor alumno: {mejor_alumno}")
````
