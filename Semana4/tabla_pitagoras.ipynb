### Descripción
Este programa construye una tabla de Pitágoras del 1 al 10 utilizando una matriz. También permite consultar el resultado de dos factores mediante funciones y validar que los números ingresados estén dentro del rango permitido.Además, permite al usuario ingresar dos factores, verificar que estén entre 1 y 10 y obtener su resultado sin realizar una nueva multiplicación, consultando directamente la matriz.

```python
tabla = []

for fila in range(1, 11):
    renglon = []
    for columna in range(1, 11):
        renglon.append(fila * columna)
    tabla.append(renglon)


def imprimir_tabla(tabla):
    for renglon in tabla:
        for numero in renglon:
            print(numero, end="\t")
        print()


def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]


imprimir_tabla(tabla)


while True:
    try:
        renglon = int(input("Ingresa el primer factor (del 1 al 10): "))

        if renglon < 1 or renglon > 10:
            print("Número fuera de rango. Ingresa un valor del 1 al 10.")
        else:
            break

    except ValueError:
        print("Entrada no válida. Debes ingresar un número entero.")


while True:
    try:
        columna = int(input("Ingresa el segundo factor (del 1 al 10): "))

        if columna < 1 or columna > 10:
            print("Número fuera de rango. Ingresa un valor del 1 al 10.")
        else:
            break

    except ValueError:
        print("Entrada no válida. Debes ingresar un número entero.")


resultado = consultar_producto(tabla, renglon, columna)

print(renglon, "x", columna, "=", resultado)
```
